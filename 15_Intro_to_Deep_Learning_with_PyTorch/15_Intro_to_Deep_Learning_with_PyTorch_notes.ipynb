```python
def show_results(model, dataloader):
    model.eval()
    iter_loader = iter(dataloader)
    for _ in range(3):
        feature, target = next(iter_loader)
        preds = model(feature)
        for p, t in zip(preds, target):
            print(f'Ground truth salary: {t.item():.3f}. Predicted salary: {p.item():.3f}.')
```

```python
# Read Data
dataset = pd.read_csv("data/salary.csv", index_col=0)

print(dataset.head())

X = dataset.iloc[:, :-1].to_numpy()
y = dataset.iloc[:, -1].to_numpy().reshape(-1, 1)

dataset = TensorDataset(torch.tensor(X, dtype=torch.float32), torch.tensor(y, dtype=torch.float32))
```

```python
num_epochs = 5
dataloader = DataLoader(dataset, batch_size=4, shuffle=True)
criterion = nn.MSELoss()
model = nn.Sequential(
    nn.Linear(in_features=4, out_features=2),
    nn.Sigmoid(),
    nn.Linear(in_features=2, out_features=1)
)
optimizer = torch.optim.SGD(model.parameters(), lr=0.001)
```

```python
# Loop over the number of epochs and the dataloader
for i in range(num_epochs):
    for data in dataloader:
        
        # Set the gradients to zero
        optimizer.zero_grad()
        
        # Run a forward pass
        feature, target = data
        prediction = model(feature)
        
        # Compute the loss
        loss = criterion(prediction, target)
        
        # Compute the gradients
        loss.backward()
        
        # Update the model's parameters
        optimizer.step()
        
show_results(model, dataloader)
```

### Learning Rate and Momentum

```python
import torch
import numpy as np
import matplotlib.pyplot as plt

def function(x):
    return x**4 + x**3 - 5*x**2

def optimize_and_plot(lr=0.01, momentum=0.0):
    x = torch.tensor(2.0, requires_grad=True)
    buffer = torch.zeros_like(x.data)
    values = []
    for i in range(15): 
        y = function(x)
        values.append((x.clone(), y.clone()))
        y.backward()    
        d_p = x.grad.data
        if momentum != 0:
            buffer.mul_(momentum).add_(d_p)
            d_p = buffer    
        x.data.add_(d_p, alpha=-lr)
        x.grad.zero_()

    x_vals = np.arange(-3, 2, 0.001)
    y_vals = function(x_vals) 
    
    plt.figure(figsize=(10, 5))
    
    # Plot optimizer steps
    plt.plot([v[0].detach().numpy() for v in values], 
            [v[1].detach().numpy() for v in values], 
            'r-X', linewidth=2, markersize=7)
            
    # Fix: Detach tensors or use .item() for text coordinates
    for i in range(15):
        plt.text(values[i][0].detach().numpy() + 0.1, 
                values[i][1].detach().numpy(), 
                f'step {i}', fontdict={'color': 'r'})
                
    plt.plot(x_vals, y_vals, linewidth=2)
    plt.grid()
    plt.tick_params(axis='both', which='major', labelsize=12)
    plt.legend(['Optimizer steps', 'Function'])
    plt.show()
```

```python
# Try a first learning rate value
lr0 = 0.001
optimize_and_plot(lr=lr0)
```

```python
# Try a second learning rate value
lr1 = 0.1
optimize_and_plot(lr=lr1)
```

```python
# Try a third learning rate value
lr2 = 0.09
optimize_and_plot(lr=lr2)
```

```python
# Try a first value for momentum
mom0 = 0.7
optimize_and_plot(momentum=mom0)
```

### Evaluating Model Performance

```python
import pickle
import numpy as np
import torch
import torch.nn.functional as F
from torch.utils.data import TensorDataset, DataLoader, random_split

with open('data/masks_dataset.pickle', 'rb') as f:
    raw_data = pickle.load(f)

images_np = raw_data[0] # Shape: (355, 32, 32, 3)
labels_np = raw_data[1] # Shape: (355,)

images_tensor = torch.tensor(images_np, dtype=torch.float32).permute(0, 3, 1, 2)

labels_tensor = torch.tensor(labels_np, dtype=torch.long)
labels_one_hot = F.one_hot(labels_tensor, num_classes=3).float()

dataset = TensorDataset(images_tensor, labels_one_hot)
train_size = int(0.8 * len(dataset))
val_size = len(dataset) - train_size

train_dataset, val_dataset = random_split(dataset, [train_size, val_size])

dataloader = DataLoader(train_dataset, batch_size=8, shuffle=True)
validationloader = DataLoader(val_dataset, batch_size=8, shuffle=False)

for batch_images, batch_labels in dataloader:
    print("Image batch shape:", batch_images.shape)  # torch.Size([32, 3, 32, 32])
    print("Label batch shape:", batch_labels.shape)  # torch.Size([32, 3])
    break
```

```python
import torch
import torch.nn as nn

class Normalize(nn.Module):
    def __init__(self, mean, std):
        super().__init__()
        self.register_buffer('mean', torch.tensor(mean).view(1, -1, 1, 1))
        self.register_buffer('std', torch.tensor(std).view(1, -1, 1, 1))

    def forward(self, x):
        return (x - self.mean) / self.std
```

```python
model = nn.Sequential(
    Normalize(
        mean=[0.51378129, 0.44999619, 0.42258797], 
        std=[0.25081163, 0.24213039, 0.24616128]
    ),
    nn.Flatten(start_dim=1, end_dim=-1),
    nn.Linear(in_features=3072, out_features=3, bias=True)
)
criterion = nn.CrossEntropyLoss()
```

```python
from torch import optim

optimizer = optim.Adam(model.parameters(), lr=0.001)
model.train()
epochs = 15
print("Starting training...\n")
for epoch in range(epochs):
    running_loss = 0.0
    for features, labels in dataloader:
        optimizer.zero_grad()
        outputs = model(features)
        
        labels_idx = torch.argmax(labels, dim=1)
        
        loss = criterion(outputs, labels_idx)
        loss.backward()
        optimizer.step()
        
        running_loss += loss.item()

print("Training completed!\n")
```

```python
# Set the model to evaluation mode
model.eval()
validation_loss = 0.0

with torch.no_grad():

    for features, labels in validationloader:
    
        outputs = model(features)
        loss = criterion(outputs, labels)

        # Sum the current loss to the validation_loss variable
        validation_loss += loss.item()
        
# Calculate the mean loss value
validation_loss_epoch = validation_loss / len(validationloader)
print(validation_loss_epoch)

# Set the model back to training mode
model.train()
```

```python
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1 import ImageGrid

def plot_errors(model, dataloader): 
    model.eval()
    mismatches = []
    
    with torch.no_grad():
        for data in dataloader:
            features, labels = data
            outputs = model(features)
            gt = labels.argmax(-1)
            pred = outputs.argmax(-1)
            for f, g, p in zip(features, gt, pred):
                if g != p:
                    mismatches.append((f, g, p))
            # Eğer yeterince hata bulduysak döngüyü durdurabiliriz
            if len(mismatches) >= 8:
                break
                
    # Eğer toplam hata sayısı 8'den azsa, grid boyutunu bulunan hata sayısına göre ayarlayalım
    num_errors = len(mismatches)
    if num_errors == 0:
        print("Harika! Model hiç hata yapmadı veya yanlış tahmin bulunamadı.")
        return
    
    # En fazla 8 tane gösterelim
    num_errors = min(num_errors, 8)
    mismatches = mismatches[:num_errors]
    
    # Satır sütun ayarı (Örn: hata sayısına göre dinamik)
    cols = 4 if num_errors >= 4 else num_errors
    rows = (num_errors + cols - 1) // cols

    fig = plt.figure(figsize=(cols * 3, rows * 3))
    grid = ImageGrid(fig, 111, 
                     nrows_ncols=(rows, cols), 
                     axes_pad=0.5, 
                     )
                     
    mapping = {0: 'No mask', 1: 'Mask', 2: 'Incorrect'}
    
    for idx, ax in enumerate(grid):
        if idx < len(mismatches):
            # Değerlerin 0-1 aralığında görünmesi için .clip() ekleyebiliriz
            img = mismatches[idx][0].permute(1, 2, 0).cpu().numpy()
            ax.imshow(np.clip(img, 0, 1))
            gt_label = mapping[mismatches[idx][1].item()]
            pred_label = mapping[mismatches[idx][2].item()]
            ax.set_title(f'GT: {gt_label} \n PRED: {pred_label}', fontsize=10)
        ax.axis('off')
        
    plt.show()
```

```python
import torchmetrics

# Create accuracy metric
metric = torchmetrics.Accuracy(task="multiclass", num_classes=3)
for features, labels in validationloader:
    outputs = model(features)
    
    # Calculate accuracy over the batch
    metric.update(outputs, labels.argmax(dim=-1))
    
# Calculate accuracy over the whole epoch
accuracy = metric.compute()
print(f"Accuracy on all data: {accuracy}")

# Reset metric for the next epoch
metric.reset()
plot_errors(model, validationloader)
```
